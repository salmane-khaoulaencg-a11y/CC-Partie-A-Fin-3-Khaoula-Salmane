## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Régression linéaire** (*Linear Regression*) est une méthode supervisée de **régression** qui cherche à expliquer et prédire une variable quantitative à partir de variables explicatives. L’idée est de représenter la relation entre les variables par une combinaison linéaire. Le processus est : **données X et y → estimation des coefficients → calcul des prédictions → évaluation des erreurs**. Le modèle s’écrit : ŷᵢ = β₀ + Σⱼ βⱼxᵢⱼ, où ŷᵢ est la valeur prédite, β₀ l’ordonnée à l’origine, βⱼ le coefficient associé à la variable j et xᵢⱼ sa valeur pour l’observation i. Les β sont des paramètres appris ; dans `LinearRegression()` utilisé ici, aucun hyperparamètre n’est fixé pour contrôler l’apprentissage. **Avantages :** méthode simple et rapide ; coefficients faciles à interpréter. **Limites :** elle représente mal les relations fortement non linéaires ; elle peut être sensible aux valeurs aberrantes et à la multicolinéarité.

### Schéma explicatif

Le schéma représente un nuage d’observations et une droite ajustée. On le lit en partant des observations : le modèle cherche la droite qui représente au mieux la tendance générale des données. La distance entre les points observés et les valeurs prédites correspond aux erreurs du modèle.

### Contextes d'utilisation

**Cas 1 — Finance :** estimer un coût ou un revenu à partir de plusieurs indicateurs financiers.  
**Cas 2 — Logistique :** prévoir le coût d’un transport selon la distance, le poids et le volume.  
**Données/prérequis :** variables quantitatives, données nettoyées et relation approximativement linéaire.  
**Peu approprié :** problème très non linéaire sans transformation des variables.  
**Métriques :** MAE, RMSE et R².

Elle calcule ensuite le **MAE**, qui mesure l’erreur absolue moyenne entre les prédictions et les valeurs réelles. ## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Ridge** (*Ridge Regression*) est une méthode de **régression** qui reprend le principe de la régression linéaire en ajoutant une pénalisation L2. Son objectif est de limiter des coefficients trop élevés et ainsi de rendre le modèle plus stable, notamment lorsque plusieurs variables sont corrélées. Le déroulement est : **standardisation → estimation des coefficients avec pénalisation → prédiction**. Le critère peut s’écrire : minβ [Σᵢ(yᵢ − β₀ − xᵢᵀβ)² + αΣⱼβⱼ²], où β₀ est l’interception, β les coefficients, xᵢ les variables explicatives et α le coefficient de régularisation. Les β sont des paramètres appris ; `alpha=1.0` est ici un hyperparamètre fixé avant l’apprentissage. **Avantages :** réduit l’effet de la multicolinéarité ; peut limiter le surapprentissage. **Limites :** ne force généralement pas les coefficients à zéro ; les performances dépendent du choix de α.

### Schéma explicatif

Le principe peut être lu comme une régression linéaire à laquelle on ajoute une « pénalité » sur la taille des coefficients. Plus α est important, plus le modèle est incité à garder des coefficients modérés.

### Contextes d'utilisation

**Cas 1 — Finance :** prédire un indicateur financier lorsque plusieurs variables sont fortement corrélées.  
**Cas 2 — Gestion :** estimer des dépenses à partir d’un grand nombre d’indicateurs budgétaires.  
**Données/prérequis :** variables numériques ; la standardisation est pertinente et est réalisée ici dans le pipeline.  
**Peu approprié :** lorsqu’on recherche une sélection stricte de variables avec des coefficients exactement nuls.  
**Métriques :** MAE, RMSE et R².

### Analyse de l'expérience Python

Le pipeline applique d’abord `StandardScaler`, puis entraîne `Ridge(alpha=1.0)` sur les données d’entraînement. Le modèle est ensuite évalué avec le MAE sur `Xr_test`. Le fichier ne contient pas la sortie numérique de Colab ; il faut donc exécuter la cellule pour obtenir la valeur. Les données étant synthétiques, le résultat sert uniquement à illustrer la méthode.


```python
# Schéma original : Ridge
import numpy as np, matplotlib.pyplot as plt
alphas=np.logspace(-3,2,60); coef=np.exp(-0.25*alphas)
plt.figure(figsize=(6,4)); plt.semilogx(alphas,coef,label="|coefficient| illustratif")
plt.xlabel("α (force de régularisation)"); plt.ylabel("Amplitude du coefficient"); plt.title("Ridge : α plus grand ⇒ coefficients davantage réduits"); plt.legend(); plt.show()
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Lasso** (*Least Absolute Shrinkage and Selection Operator*) appartient à la famille de **régression**. Comme Ridge, il ajoute une régularisation, mais utilise ici une pénalisation L1. Son intérêt principal est de réduire certains coefficients jusqu’à exactement zéro, ce qui peut conduire à une sélection automatique de variables. Le principe est : **standardisation → estimation pénalisée → prédiction**. Le critère s’écrit : minβ [Σᵢ(yᵢ − β₀ − xᵢᵀβ)² + αΣⱼ|βⱼ|], où α contrôle la force de la pénalisation et β représente les coefficients appris. `alpha` est donc un hyperparamètre. **Avantages :** peut sélectionner des variables ; limite le surapprentissage. **Limites :** peut être instable lorsque les variables sont fortement corrélées ; un α trop élevé peut supprimer des variables utiles.

### Schéma explicatif

Le schéma doit être interprété comme une régression à laquelle on ajoute une pénalité basée sur la valeur absolue des coefficients. Cette pénalité peut pousser certains coefficients jusqu’à zéro, d’où l’effet de sélection de variables.

### Contextes d'utilisation

**Cas 1 — Finance :** sélectionner les indicateurs les plus utiles pour expliquer un risque financier.  
**Cas 2 — Marketing :** prédire une réponse commerciale à partir de nombreuses variables clients.  
**Données/prérequis :** variables numériques ; standardisation recommandée.  
**Peu approprié :** lorsque des variables très corrélées doivent toutes être conservées de façon stable.  
**Métriques :** MAE, RMSE et R².

### Analyse de l'expérience Python

Le modèle `Lasso(alpha=0.05)` est utilisé dans un pipeline avec standardisation. Il apprend sur `Xr_ent` et `yr_ent`, puis les prédictions sont évaluées avec le MAE sur `Xr_test`. La valeur numérique n’est pas enregistrée dans le notebook et doit être obtenue après exécution. L’expérience repose sur des données synthétiques.


```python
# Schéma original : Lasso
import numpy as np, matplotlib.pyplot as plt
a=np.linspace(0,2,80); c=np.maximum(0,1-a)
plt.figure(figsize=(6,4)); plt.plot(a,c,label="coefficient illustratif"); plt.axhline(0,linewidth=1)
plt.xlabel("α"); plt.ylabel("Coefficient"); plt.title("Lasso : certains coefficients peuvent devenir nuls"); plt.legend(); plt.show()
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Régression polynomiale** (*Polynomial Regression*) est une méthode de **régression** permettant de représenter des relations non linéaires entre les variables et la cible en ajoutant des puissances et interactions des variables. Le principe est : **variables d’entrée → création de caractéristiques polynomiales → régression linéaire → prédiction**. Par exemple, un modèle peut contenir x, x² et x³. Les coefficients β sont appris pendant l’entraînement, tandis que le **degré** du polynôme est un hyperparamètre choisi avant l’apprentissage. **Avantages :** permet de modéliser certaines courbures ; reste relativement simple à interpréter pour un petit degré. **Limites :** un degré trop élevé peut provoquer du surapprentissage ; le nombre de variables peut augmenter rapidement.

### Schéma explicatif

Le schéma montre que les variables initiales sont transformées en nouvelles caractéristiques, par exemple x² ou x³, avant d’être transmises à une régression. La relation finale peut donc être courbe même si le modèle appliqué après transformation est linéaire par rapport à ses coefficients.

### Contextes d'utilisation

**Cas 1 — Finance :** modéliser une relation courbe entre un indicateur financier et une variable économique.  
**Cas 2 — Logistique :** estimer un coût qui augmente de manière non linéaire avec la distance ou le volume.  
**Données/prérequis :** variables numériques et choix raisonnable du degré.  
**Peu approprié :** très nombreuses variables avec un degré élevé, car le nombre de caractéristiques peut exploser.  
**Métriques :** MAE, RMSE et R².

### Analyse de l'expérience Python

L’expérience montre comment l’ajout de termes non linéaires permet d’obtenir une relation plus flexible. La valeur de performance doit être relevée après exécution si elle est affichée ; aucune valeur absente du notebook n’est ajoutée ici. Les données sont synthétiques.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Arbre de régression** (*Decision Tree Regressor*) est un algorithme supervisé de **régression**. Il divise progressivement les observations en groupes à l’aide de règles sur les variables, jusqu’à obtenir des feuilles contenant des prédictions. Le processus est : **choix d’une coupure → séparation des données → répétition → valeur prédite dans la feuille**. Pour la régression, une coupure cherche généralement à réduire l’erreur, par exemple la somme des carrés des résidus. Les paramètres de l’arbre sont appris à partir des données ; des hyperparamètres comme `max_depth` limitent sa complexité. **Avantages :** capture des relations non linéaires ; ne nécessite généralement pas de standardisation. **Limites :** risque de surapprentissage ; petites variations des données peuvent modifier fortement l’arbre.

### Schéma explicatif

Le schéma se lit de la racine vers les feuilles. À chaque nœud, une règle sépare les observations. En arrivant à une feuille, le modèle fournit une valeur numérique correspondant à la prédiction pour les observations qui y sont arrivées.

### Contextes d'utilisation

**Cas 1 — Finance :** estimer une perte ou un montant financier à partir de caractéristiques d’un client.  
**Cas 2 — Logistique :** prévoir un coût ou un délai selon les caractéristiques d’une expédition.  
**Données/prérequis :** données tabulaires, avec encodage des variables catégorielles si nécessaire.  
**Peu approprié :** lorsqu’une très grande stabilité du modèle est indispensable et qu’un arbre unique varie trop selon l’échantillon.  
**Métriques :** MAE, RMSE et R².

### Analyse de l'expérience Python

La cellule entraîne un `DecisionTreeRegressor` sur les données de régression puis produit des prédictions sur le jeu de test. Le résultat numérique doit être lu après exécution. L’intérêt de l’expérience est de montrer qu’un arbre peut apprendre des relations non linéaires sans standardisation. Les données restent synthétiques.


```python
# Schéma original : arbre de régression
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
X,y=make_regression(n_samples=80,n_features=1,noise=10,random_state=42)
m=DecisionTreeRegressor(max_depth=2,random_state=42).fit(X,y)
plt.figure(figsize=(10,5)); plot_tree(m,filled=False,feature_names=["X"]); plt.title("Arbre de régression : partitions successives"); plt.show()
```

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Présentation

**Forêt aléatoire — régression** (*Random Forest Regressor*) est une méthode d’ensemble de **régression** qui combine plusieurs arbres de décision. Chaque arbre apprend sur une variation des données et des variables, puis leurs prédictions sont agrégées. Le déroulement est : **construction de plusieurs arbres → prédiction de chaque arbre → moyenne des prédictions**. On peut représenter la sortie par ŷ = (1/T)Σₜ fₜ(x), où T est le nombre d’arbres et fₜ la prédiction de l’arbre t. `n_estimators` est un hyperparamètre. **Avantages :** bonne capacité à modéliser des relations complexes ; réduit la variance d’un arbre seul. **Limites :** plus coûteuse en calcul et en mémoire qu’un arbre unique ; interprétation moins directe.

### Schéma explicatif

Plusieurs arbres reçoivent les mêmes variables d’entrée mais apprennent des modèles différents. Leurs prédictions numériques sont ensuite moyennées pour produire une seule estimation finale.

### Contextes d'utilisation

**Cas 1 — Finance :** prévoir un montant de risque ou de perte.  
**Cas 2 — Logistique :** estimer un délai ou un coût de livraison.  
**Données/prérequis :** données tabulaires ; standardisation généralement inutile.  
**Peu approprié :** modèle devant être expliqué sous la forme d’une règle unique et très courte.  
**Métriques :** MAE, RMSE et R².

### Schéma explicatif

Le schéma se lit étape par étape : les premières observations sont traitées, les erreurs sont repérées, puis les observations difficiles reçoivent plus de poids. Le modèle suivant essaie alors de mieux les classer.

**K-means** (*K-Means Clustering*) est une méthode non supervisée de **clustering**. Son objectif est de répartir les observations en K groupes autour de centres appelés centroïdes. ### Schéma explicatif

On commence avec plusieurs centroïdes. Les points sont attribués au centroïde le plus proche, puis chaque centroïde est déplacé vers le centre de son groupe. Ces deux étapes sont répétées jusqu’à stabilisation.

Les estimations des récompenses sont des paramètres appris ; la stratégie de sélection est contrôlée par des hyperparamètres selon l’algorithme choisi. **Avantages :** cadre simple pour les décisions séquentielles ; permet d’apprendre sans connaître à l’avance la meilleure action. **Limites :** compromis exploration/exploitation ; hypothèses parfois trop simples pour un environnement avec états complexes.

### Schéma explicatif

L’agent choisit un « bras », reçoit une récompense puis met à jour son estimation. Il recommence et utilise progressivement les informations accumulées pour améliorer ses choix.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Présentation

**Epsilon-greedy** est une stratégie de décision utilisée dans les **bandits et l’apprentissage par renforcement** pour gérer le compromis exploration/exploitation. Avec une probabilité ε, l’agent explore en choisissant une action de manière non déterministe ; avec une probabilité 1−ε, il exploite l’action ayant la meilleure estimation actuelle. Le fonctionnement est : **observer les estimations → tirer au sort exploration/exploitation → recevoir la récompense → mettre à jour l’estimation**. ε est un hyperparamètre de la stratégie. **Avantages :** très simple à mettre en œuvre ; garantit une exploration minimale lorsque ε>0. **Limites :** les explorations peuvent être inefficaces ; un ε fixe peut rester trop élevé ou trop faible selon le moment de l’apprentissage.

### Schéma explicatif

À chaque décision, deux chemins sont possibles : « explorer » avec probabilité ε ou « exploiter » avec probabilité 1−ε. Après l’action, la récompense obtenue sert à améliorer l’estimation de l’action choisie.

### Contextes d'utilisation

**Cas 1 — Marketing :** tester différentes offres tout en privilégiant progressivement la plus performante.  
**Cas 2 — Commerce :** choisir entre plusieurs actions commerciales avec exploration contrôlée.  
**Données/prérequis :** environnement donnant une récompense après chaque action.  
**Peu approprié :** situation où les décisions ne sont pas répétées et où aucune interaction n’est possible.  
**Métriques :** récompense cumulée, taux de réussite et regret.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Présentation

**Q-learning** est un algorithme d’apprentissage par **renforcement** qui apprend la valeur d’une action dans un état donné. Il cherche à estimer Q(s,a), c’est-à-dire la récompense cumulée attendue lorsqu’on choisit l’action a dans l’état s et qu’on suit ensuite une stratégie optimale. La mise à jour fondamentale est : Q(s,a) ← Q(s,a) + α[r + γ maxₐ′Q(s′,a′) − Q(s,a)], où α est le taux d’apprentissage, γ le facteur d’actualisation, r la récompense et s′ le nouvel état. **Avantages :** simple pour des espaces d’états/actions discrets ; apprend une politique sans modèle explicite de l’environnement. **Limites :** table Q difficile à utiliser lorsque l’espace devient très grand ; apprentissage dépendant de l’exploration.

### Schéma explicatif

L’agent observe l’état s, choisit une action a et reçoit une récompense r ainsi qu’un nouvel état s′. Il utilise ces informations pour mettre à jour Q(s,a), puis recommence.

### Contextes d'utilisation

**Cas 1 — Logistique :** apprendre des décisions successives de déplacement ou d’affectation dans un environnement simplifié.  
**Cas 2 — Gestion :** apprendre une politique de décision lorsque les actions produisent des récompenses à long terme.  
**Données/prérequis :** environnement interactif et espaces d’états/actions suffisamment petits ou discrétisés.  
**Peu approprié :** très grand espace continu sans représentation adaptée.  
**Métriques :** récompense cumulée, taux de réussite et stabilité entre graines.

### Analyse de l'expérience Python

L’expérience met à jour une table Q au fur et à mesure des interactions avec l’environnement. Le modèle ne reçoit pas une colonne cible fixe comme en apprentissage supervisé. Les résultats doivent donc être interprétés à travers la récompense obtenue et l’évolution de la politique. L’environnement de démonstration ne constitue pas un dataset économique réel.


```python
# Schéma original : Q-learning
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(7,4)); ax.axis("off")
labels=[("État s",.12,.65),("Action a",.38,.65),("Récompense r",.64,.65),("Nouvel état s'",.88,.65),("Mise à jour Q",.5,.2)]
for s,x,y in labels: ax.text(x,y,s,ha="center",bbox=dict(boxstyle="round"))
for a,b in [((.18,.65),(.32,.65)),((.44,.65),(.58,.65)),((.70,.65),(.82,.65))]: ax.annotate("",xy=b,xytext=a,arrowprops=dict(arrowstyle="->"))
ax.annotate("",xy=(.5,.25),xytext=(.88,.58),arrowprops=dict(arrowstyle="->")); ax.annotate("",xy=(.12,.58),xytext=(.5,.25),arrowprops=dict(arrowstyle="->"))
ax.set_title("Q-learning : mise à jour avec le meilleur futur possible"); plt.show()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Présentation

**SARSA** (*State–Action–Reward–State–Action*) est un algorithme d’apprentissage par **renforcement** qui apprend également une fonction Q. Sa particularité est que la mise à jour utilise l’action suivante effectivement choisie par la politique : Q(s,a) ← Q(s,a) + α[r + γQ(s′,a′) − Q(s,a)]. Ici, α est le taux d’apprentissage, γ le facteur d’actualisation, r la récompense et a′ l’action suivante réellement sélectionnée. Le processus est : **état → action → récompense et nouvel état → nouvelle action → mise à jour**. **Avantages :** simple pour des espaces discrets ; prend en compte le comportement réel de la stratégie de choix. **Limites :** dépend de la stratégie d’exploration ; peu adapté aux très grands espaces sans approximation.

### Schéma explicatif

Contrairement à une mise à jour qui regarde directement la meilleure action possible, SARSA suit la trajectoire effectivement choisie : s, a, r, s′, a′. La valeur de Q(s,a) est ensuite ajustée avec cette information.

### Contextes d'utilisation

**Cas 1 — Logistique :** apprendre une politique de décisions séquentielles dans un environnement simplifié.  
**Cas 2 — Gestion :** optimiser progressivement une suite de décisions lorsque les récompenses sont différées.  
**Données/prérequis :** environnement interactif et états/actions suffisamment représentables.  
**Peu approprié :** espace d’états très vaste ou fortement continu sans fonction d’approximation.  
**Métriques :** récompense cumulée, réussite et stabilité entre plusieurs graines.

### Analyse de l'expérience Python

La cellule met à jour Q à partir de la transition complète observée : état, action, récompense, nouvel état et action suivante. Le comportement appris doit être évalué par les récompenses obtenues dans l’environnement. Il faut distinguer cette simulation interactive d’un jeu de données réel fixe.


```python
# Schéma original : SARSA
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(7,4)); ax.axis("off")
for s,x in [("s,a",.15),("r,s'",.38),("a'",.62),("Q(s,a) ← cible",.85)]:
    ax.text(x,.55,s,ha="center",bbox=dict(boxstyle="round"))
for x1,x2 in [(.21,.32),(.44,.56),(.68,.79)]: ax.annotate("",xy=(x2,.55),xytext=(x1,.55),arrowprops=dict(arrowstyle="->"))
ax.text(.5,.18,"La cible utilise l'action réellement choisie a' : Q(s',a')",ha="center")
ax.set_title("SARSA : mise à jour on-policy"); plt.show()
```
